### In Disk Storage

```python
import chromadb
import uuid

client = chromadb.PersistentClient("./croma_data")

collection = client.get_or_create_collection("polices")

with open("northstar_delivery_policy.txt", "r", encoding= "utf-8") as f:
    polices: list[str] = f.read().split(".")

collection.add(
    ids = [str(uuid.uuid4()) for _ in polices],
    documents=polices,
    metadatas=[{"line": line} for line in range(len(polices))]

)
```

```python
print(collection.peek(3))
```

```python
print("Total records:", collection.count())
```

```python
print(collection.peek(5))
```

```python
result = collection.peek(1)

embedding = result["embeddings"][0]

print("Embedding dimension:", len(embedding))
```

```python
results = collection.query(
    query_texts=["what are the parcel acceptance poloicy"],
    n_results=5
)

for document in results["documents"][0]:
    print(document)
    print("--------------------")
```
